```python
from pyspark.sql.functions import expr
df2 = df.select("name",expr("country AS my_country"))
display(df2)
```

```python
from pyspark.sql.functions import expr
df2 = df.selectExpr("name","country as my_country")
display(df2)
```

```python
from pyspark.sql.functions import expr
df2 = df.select('name',expr("concat('city','_','country')").alias('location'))

display(df2)
```

```python
%sql
select name,concat(city,"_",country) as location from samples.bakehouse.sales_franchises
```

```python
df2 = df.withColumn('location',expr("concat(city,'_',country)"))

display(df2)
```

```python
%sql
select *,concat(city,'_',country) as location from samples.bakehouse.sales_franchises
```

```python
from pyspark.sql.functions import lit
df2 = df.withColumn('one',lit(1))

display(df2)
```
