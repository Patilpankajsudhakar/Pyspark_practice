```python
# Reading corrupted CSV data
df_csv = spark.read.option("header",'true').\
        format('csv').\
        load("/Volumes/my_files/my_schema/csv_files/corrupated_csv_file.csv")
display(df_csv)
```

```python
# specify the schema
schema_str = "CustomerID string, Name string, Age INT, Email string, Phone DOUBLE, City string, Salary DOUBLE, JoinDate Date,_corrupt_record string"

df_csv = spark.read.schema(schema_str).\
        option("header",'true').\
        option("mode","PERMISSIVE").\
        format('csv').\
        load("/Volumes/my_files/my_schema/csv_files/corrupated_csv_file.csv")
display(df_csv)
```

```python
# option 2 - badRecoard path

df_csv = spark.read.schema(schema_str)\
        .option("header",'true')\
        .format('csv')\
        .option("badRecordsPath","/Volumes/my_files/my_schema/csv_files/bad_records_path")\
        .load("/Volumes/my_files/my_schema/csv_files/corrupated_csv_file.csv")
display(df_csv)
```
