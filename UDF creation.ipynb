```python
# user defined function

data = [
    ("Order ID: 54863",),
    ("Invoice 98321 processed",),
    ("No number here",),
    ("Ref-45A128",)
]

df = spark.createDataFrame(data,["text_col"])
df.createOrReplaceTempView("orders")
display(df)
```

```python
import re

# define python fuction for extraction

def extract_number(text):
    if text is None:
        return None
    match = re.search(r"\d+", text) 
    return match.group(0) if match else None
```

```python
# Register the UDF
spark.udf.register("extractNumberUDF",extract_number)
```

```python
%scala

import org.apache.spark.sql.SparkSession
import scala.util.matching.Regex

// scala function to extract first numeric sequence form text

def extractNumber(text: String): String = {
    if (text == null) return null

    val pattern : Regex = "\\d+".r
    pattern.findFirstIn(text) match {
        case Some(value) => value
        case None => null
    }
}

// register as udf

spark.udf.register("extractNumberUDF",extractNumber)
```

```python
# used

from pyspark.sql.functions import col
from pyspark.sql.functions import expr

df2 = df.withColumn("extracted_number", expr("extractNumberUDF(text_col)")).show()

display(df2)
```
